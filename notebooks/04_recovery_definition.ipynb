# 04 — Recovery definition

Mobility Disruption Score combines radius, displacement, and daily-distance deviations.
A user is recovered on the first of two consecutive days inside a 20% band of their own baseline radius, displacement, and daily distance.

```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT))

from src.features import compute_baseline_features, compute_daily_metrics, estimate_home
from src.ingest import attach_disaster_clock, filter_events, filter_users, load_events, load_raw
from src.plots import mean_mds_by_day, plot_mds_over_time
from src.recovery import attach_disruption, label_recovery

tweets = attach_disaster_clock(filter_events(load_raw(), ["14_Napa"]), load_events())
filtered, _ = filter_users(tweets)
home = estimate_home(filtered)
daily = compute_daily_metrics(filtered, home)
baseline = compute_baseline_features(filtered, daily)
disrupted = attach_disruption(daily, baseline)
labels = label_recovery(disrupted)
print(labels["event_observed"].value_counts())
labels.head()
```

```python
mds = mean_mds_by_day(disrupted)
fig = plot_mds_over_time(mds, title="Napa — mean MDS by day")
fig.show()
```
