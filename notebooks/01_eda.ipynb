# 01 — Exploratory data analysis

Source: Wang & Taylor (2016) geotagged tweets during 15 natural disasters.

Start with **13_Napa** (2014 South Napa earthquake, 03:20 PDT on 24 August). Longitude is anonymized; distances within an event are still valid.

```python
import sys
from pathlib import Path

import pandas as pd

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT))

from src.ingest import attach_disaster_clock, event_time_coverage, load_events, load_raw

events = load_events()
events
```

```python
raw = load_raw()
print(raw.shape)
raw.head()
```

```python
print(raw.groupby("disaster").agg(n_obs=("time", "size"), n_users=("user_anon", "nunique")))
```

```python
clocked = attach_disaster_clock(raw, events)
coverage = event_time_coverage(clocked)
coverage
```

```python
napa = clocked.loc[clocked["disaster"] == "13_Napa"]
print(napa["time"].min(), napa["time"].max())
print("users", napa["user_anon"].nunique(), "obs", len(napa))
napa["day_relative"].value_counts().sort_index().head(20)
```
